### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja4.head()
```

#1. Análise do faturamento



```python
#Calculando o faturamento total de cada loja usado afunção sum()
#Para somar os valores da coluna preço
faturamento = loja["Preço"].sum()
faturamento2 = loja2["Preço"].sum()
faturamento3 = loja3["Preço"].sum()
faturamento4 = loja4["Preço"].sum()

#Imprimindo o faturamento de cada loja
print(f"Faturamento Loja: R$ {faturamento}")
print(f"Faturamento Loja2: R$ {faturamento2}")
print(f"Faturamento Loja3: R$ {faturamento3}")
print(f"Faturamento Loja4: R$ {faturamento4}")
```

```python
#importando biblioteca
import matplotlib.pyplot as plt

#Definindo os dados que quero mostrar
faturamento = [faturamento, faturamento2, faturamento3, faturamento4]
lojas = ["Loja 1", "Loja 2", "Loja 3", "Loja 4"]

#Tipo de grafico
#Para comparar os valores escolhi o de barras
plt.bar(lojas, faturamento)

#Personalizando com titulo e rótulos:
plt.title("Faturamento por Loja")
plt.xlabel("Lojas")
plt.ylabel("Faturamento (R$)")

#Mostrando o gráfico:
plt.show()

```

# 2. Vendas por Categoria

```python
#Agrupando as categotias e somando os preços de cada uma
vendas_loja = loja.groupby("Categoria do Produto")["Preço"].sum()
vendas_loja2 = loja2.groupby("Categoria do Produto")["Preço"].sum()
vendas_loja3 = loja3.groupby("Categoria do Produto")["Preço"].sum()
vendas_loja4 = loja4.groupby("Categoria do Produto")["Preço"].sum()


import matplotlib.pyplot as plt

# Loja 1
vendas_categoria1 = loja.groupby("Categoria do Produto")["Preço"].sum()
plt.figure(figsize=(8, 8)) # aumenta o tamanho do gráfico
vendas_categoria1.plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribuição das Vendas por Categoria - Loja 1")
plt.ylabel("")  # tira o rótulo automático do eixo
plt.show()

# Loja 2
vendas_categoria2 = loja2.groupby("Categoria do Produto")["Preço"].sum()
plt.figure(figsize=(8, 8)) # aumenta o tamanho do gráfico
vendas_categoria2.plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribuição das Vendas por Categoria - Loja 2")
plt.ylabel("")
plt.show()

# Loja 3
vendas_categoria3 = loja3.groupby("Categoria do Produto")["Preço"].sum()
plt.figure(figsize=(8, 8)) # aumenta o tamanho do gráfico
vendas_categoria3.plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribuição das Vendas por Categoria - Loja 3")
plt.ylabel("")
plt.show()

# Loja 4
vendas_categoria4 = loja4.groupby("Categoria do Produto")["Preço"].sum()
plt.figure(figsize=(8, 8)) # aumenta o tamanho do gráfico
vendas_categoria4.plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribuição das Vendas por Categoria - Loja 4")
plt.ylabel("")
plt.show()

```

# 3. Média de Avaliação das Lojas


```python
# Média das avaliações
avaliacao1 = loja["Avaliação da compra"].mean().round(2)
avaliacao2 = loja2["Avaliação da compra"].mean().round(2)
avaliacao3 = loja3["Avaliação da compra"].mean().round(2)
avaliacao4 = loja4["Avaliação da compra"].mean().round(2)

# Imprimindo o resultado
print(f"Avaliação média Loja1: {avaliacao1}")
print(f"Avaliação média Loja2: {avaliacao2}")
print(f"Avaliação média Loja3: {avaliacao3}")
print(f"Avaliação média Loja4: {avaliacao4}")

```

```python
import matplotlib.pyplot as plt


# Dados para o gráfico
lojas = ["Loja 1", "Loja 2", "Loja 3", "Loja 4"]
avaliacoes = [avaliacao1, avaliacao2, avaliacao3, avaliacao4]

# Criando gráfico de linhas
plt.plot(lojas, avaliacoes, marker="o", linestyle="-", color="green")

# Personalizando
plt.title("Média das Avaliações por Loja")
plt.xlabel("Lojas")
plt.ylabel("Avaliação Média")
plt.ylim(0, 5)  # se as avaliações forem de 0 a 5
plt.grid(True)

# Mostrar gráfico
plt.show()
```

# 4. Produtos Mais e Menos Vendidos

```python
import pandas as pd

# Calculando mais e menos vendidos por loja
vendas_produtos1 = loja.groupby("Produto")["Preço"].count()
mais_vendido1 = vendas_produtos1.idxmax()
menos_vendido1 = vendas_produtos1.idxmin()

vendas_produtos2 = loja2.groupby("Produto")["Preço"].count()
mais_vendido2 = vendas_produtos2.idxmax()
menos_vendido2 = vendas_produtos2.idxmin()

vendas_produtos3 = loja3.groupby("Produto")["Preço"].count()
mais_vendido3 = vendas_produtos3.idxmax()
menos_vendido3 = vendas_produtos3.idxmin()

vendas_produtos4 = loja4.groupby("Produto")["Preço"].count()
mais_vendido4 = vendas_produtos4.idxmax()
menos_vendido4 = vendas_produtos4.idxmin()

# Criando tabela com os resultados
dados = {
    "Loja": ["Loja 1", "Loja 2", "Loja 3", "Loja 4"],
    "Mais Vendido": [mais_vendido1, mais_vendido2, mais_vendido3, mais_vendido4],
    "Menos Vendido": [menos_vendido1, menos_vendido2, menos_vendido3, menos_vendido4]
}

tabela_vendas = pd.DataFrame(dados)

# Mostrando a tabela
print(tabela_vendas)

```

# 5. Frete Médio por Loja

```python
#Calculando a média do frete com mean()
#Arredondando para 2 casas decimal o resultado da média
frete = loja['Frete'].mean().round(2)
frete2 = loja2['Frete'].mean().round(2)
frete3 = loja3['Frete'].mean().round(2)
frete4 = loja4['Frete'].mean().round(2)

#imprimindo a média de frete por loja
print(f"Frete médio loja1: {frete}")
print(f"Frete médio loja2: {frete2}")
print(f"Frete médio loja3: {frete3}")
print(f"Frete médio loja4: {frete4}")
```
